```python
def fn_prime(ele):
  if ele>1:
    for i in range(2,ele):
      if ele%i==0:
        print(ele,"is not prime")
        break
    else:
     print(ele,"is prime")
  else:
    print(ele,"is not prime")
```

```python
fn_prime(ele=7)
fn_prime(ele=27)
```

```python
a=[2,5,7,25,32]
for ele in a:
  fn_prime(ele)
```
