```python
import pandas as pd
# Create a DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob','lean','holly'],
    'Age': [25, 30,42,31],
    'city':['newyork','london','sydney','newcastle']
})
# Filter rows where Age > 28 (keep both Name and Age)
filtered_df = df[df['Age'] > 28]
# Save to CSV
filtered_df.to_csv('people_over_28.csv', index=False)

from google.colab import drive
drive.mount('/content/drive')



```
