```python
def fn1(fn_a,fn_b):
  print(f"fn_a={fn_a}")
  print(f"fn_b={fn_b}")
```

```python
def fn2(fn_a,fn_b=1):
  print(f"fn_a={fn_a}")
  print(f"fn_b={fn_b}")
```

```python
a=88
b=10
```

```python
fn1(a,b)
```

```python
fn1(a)
```

```python
fn1(b,a)
```

```python
fn1(fn_a=a,fn_b=b)
```

```python
fn1(fn_b=b,fn_a=a)
```

```python
fn2(fn_a=a)
```

```python
fn2(a,b)
```

```python
fn2(a)
```

```python
fn2(fn_b=10)
```

```python
def fn_args(*args):
  print(*args)
  length=len(args)
  print(length)
  print(type(args))
```

```python
fn_args(1)
```

```python
fn_args(2,5)
```

```python
fn_args()
```

```python
fn_args(1,2,3,4,5,6,7,8,9)
```

```python
def named_param_fn(**kwargs):
  print(kwargs)
  l=len(kwargs)
  print("length=",l)
  print(type(kwargs))
```

```python
named_param_fn(a=7)
```
